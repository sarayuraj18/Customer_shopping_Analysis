```python
import pandas as pd

# Step 1: Load dataset (change path if needed)
df = pd.read_csv(r"C:\Users\kalya\Downloads\customer_shopping_behavior.csv")


# Step 2: Clean column names
df.columns = df.columns.str.replace(' ', '_')
df.columns = df.columns.str.replace('(', '')
df.columns = df.columns.str.replace(')', '')

# Step 3: Handle missing values
df['Review_Rating'] = df['Review_Rating'].fillna(df['Review_Rating'].mean())

# Step 4: Feature Engineering (Age Group)
df['Age_Group'] = pd.cut(df['Age'],
                        bins=[18,25,35,50,70],
                        labels=['18-25','25-35','35-50','50+'])

# Step 5: Save cleaned file
df.to_csv("cleaned_data.csv", index=False)

print("CSV created successfully ✅")
```
